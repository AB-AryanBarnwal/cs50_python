```python
#set datatype

students = [
    {"name": "Hermione", "house": "Griffindor"},
    {"name": "Harry", "house": "Griffindor"},
    {"name": "Ron", "house": "Griffindor"},
    {"name": "Draco", "house": "Slytherin"},
    {"name": "Padma", "house": "Ravenclaw"}
]

house = set()
for student in students:
    house.add(student["house"])


for h in sorted(house):
    print(h)
```

```python
#global and local variable
#you can read global variable inside a function but not edit it unless global keyword is written in function

balance = 0

def main():
    print(balance)
    deposit(100)
    withdraw(50)
    print(balance)

def deposit(n):
    global balance
    balance+=n

def withdraw(n):
    global balance
    balance-=n

if __name__ == "__main__":
    main()
```

```python
class Account:
    def __init__(self):
        self._balance = 0

    @property
    def balance(self):
        return self._balance
    
    def deposit(self, n):
        self._balance += n

    def withdraw(self, n):
        self._balance-=n

def main():
    account=Account()
    print("Balance:", account.balance)
    account.deposit(100)
    account.withdraw(50)
    print("Balance:", account.balance)

main()
```

```python
#docstrings

def meow(n: int) -> str:
    """
    Meow n times.

    : param n: Number of times to meow
    :type n: int
    :raise TypeError: If n is not an int
    :return: A string of n meows, one per line
    :rtype: str
    """
    return "meow\n" * n

number = int(input("Number: "))
meows: str = meow(number)
print(meows, end="")
```

```python
def main():
    yell("This", "is", "CS50")

def yell(*words):
    uppercased = []
    for word in words:
        uppercased.append(word.upper())
    print(*uppercased)

main()
```

```python
def main():
    yell("This", "is", "CS50")

def yell(*words):
    uppercased = map(str.upper, words)
    print(*uppercased)

main()
```

```python
#list comprehension

def main():
    yell("This", "is", "CS50")

def yell(*words):
    uppercased = [word.upper() for word in words]
    print(*uppercased)

main()
```

```python
students = [
    {"name": "Hermione", "house": "Griffindor"},
    {"name": "Harry", "house": "Griffindor"},
    {"name": "Ron", "house": "Griffindor"},
    {"name": "Draco", "house": "Slytherin"},
    {"name": "Padma", "house": "Ravenclaw"}
]

griffindors = [
    student["name"] for student in students if student["house"] == "Griffindor"
]

print(*griffindors)
```

```python
#filter

students = [
    {"name": "Hermione", "house": "Griffindor"},
    {"name": "Harry", "house": "Griffindor"},
    {"name": "Ron", "house": "Griffindor"},
    {"name": "Draco", "house": "Slytherin"},
    {"name": "Padma", "house": "Ravenclaw"}
]

def is_gryffindor(s):
    return s['house'] == "Gryffindor"

gryffindors = filter(is_gryffindor, students)

print(*griffindors)
```

```python
#dictionary comprehension

students = ["Hermione", "Ron", "Harry"]

gryffindors = {student: "Gryffindor" for student in students}

print(gryffindors)
```

```python
#generators

def main():
    n = int(input("What's n? "))
    for s in sheep(n):
        print(s)
    
def sheep(n):
    for i in range(n):
        yield "sheep " * i

main()
```

```python
def factorial(n):
    if n == 1 or n == 0:
        return 1
    return n * factorial(n-1)


result = factorial(int(input("No. ")))
print(result)
```
